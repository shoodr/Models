```python
import pandas as pd
import numpy as np
from sklearn import linear_model
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('/content/weather_clean.csv')
```

```python
df
```

```python
model = linear_model.LinearRegression()
```

```python
model.fit(df[['temperature_2m_max (°C)','temperature_2m_min (°C)','uv_index_max ()','wind_speed_10m_max (km/h)',]],df[['weather_code (wmo code)']])
```

```python
x=df[['temperature_2m_max (°C)','temperature_2m_min (°C)','uv_index_max ()','wind_speed_10m_max (km/h)',]]
y=df[['weather_code (wmo code)']]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_train,y_train)
```
