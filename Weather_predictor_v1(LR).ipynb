```python
import pandas as pd
import seaborn as sns

from sklearn import linear_model
import numpy as np







```

```python
df = pd.read_csv('/content/rain_probability_daily_avg (1).csv')
```

```python
df
```

```python
rng = np.random.default_rng(42)          # fixed seed so the split is repeatable
idx = rng.permutation(len(df))           # shuffled row positions
cut = int(0.8 * len(df))
```

```python
train = df.iloc[idx[:cut]]
test  = df.iloc[idx[cut:]]
```

```python
df["day"] = pd.to_datetime(df["day"])
df["day_num"] = (df["day"] - df["day"].min()).dt.days

```

```python
sns.lmplot(x='day_num',y='rain_probability',data=df)
```

```python
model =linear_model.LinearRegression()
```

```python
model.fit(df[['day_num']],df[['rain_probability']])
```

```python
model.score(df[['day_num']],df[['rain_probability']])
```
